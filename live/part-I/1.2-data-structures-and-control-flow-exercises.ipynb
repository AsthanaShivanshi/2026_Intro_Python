## Exercise 8: Default arguments and returning two values

Write a function `summarise(values, scale=1.0)` that multiplies every value by `scale` and
returns **two** things: the total and the mean.

1. Call it on `[6.1, 7.4, 5.9, 8.2, 6.8]` with the default scale, unpacking the result into
   `total_celsius` and `mean_celsius`.
2. Call it again with `scale=2.0` passed by keyword, and print only the mean.
3. Call it once more and store the result in a single name. Print its `type` — what did you get?

````python
def summarise(values, scale=1.0):
    for i in range(len(values)):
        values[i] *= scale
        total = sum(values)
        average = sum(values) / len(values)

    return total, average
````

## Exercise 9: A variable number of arguments

Write a function `warmest(*temps_celsius)` (no type hints) that returns the highest value among

### Exercise 15: Reject impossible temperatures

Write `to_kelvin(temp_celsius)` that raises a `ValueError` when the input is below absolute zero
(−273.15 °C) and otherwise returns the temperature in kelvin. Give it a docstring and a type hint.

Loop over `[20.0, -300.0, 0.0]`, calling it inside a `try`/`except` so that the invalid value is
reported without stopping the loop.

````python
def to_kelvin(temp_celsius:float) -> float:
    if temp_celsius < -273.15:
        raise ValueError("This temp is impossible")
    return temp_celsius + 273.15
````

````python
temp_list= [20.0, -300.0, 0.0]
````

### C: functions

**Q9) Write a function that converts a planetary mass to Earth masses.**

Jupiter is about 1898 x 10^24 kg, which is about 318 Earth masses:

$$M_{\text{Jupiter}} \approx 1898 \times 10^{24}\,\text{kg} \approx 318\,M_{\text{Earth}}$$

Call it `to_earth_masses(mass_1e24kg)` and check it against Jupiter — you should get
317.92294807370183.

````python
# Q9: write to_earth_masses here, then test it on Jupiter


def earth_mass(planet):
    return masses[planet] / masses["earth"] 
````

````python
earth_mass("jupiter")
````

**Q10) Now write a single function that converts to either Earth masses or Jupiter
masses, depending on a keyword argument.**

Call it `to_planet_masses(mass_1e24kg, reference="jupiter")`. Converting Jupiter's own mass should
give 317.923 with `reference="earth"` and exactly 1.0 with `reference="jupiter"`.

With the default argument, one function covers both conversions. Jupiter in Jupiter masses has
to come out as exactly 1.0; any other value means the function divides by the wrong mass.

````python
# Q10: write to_planet_masses here, then test both references on Jupiter
````

````python
def to_planet_mass(planet, reference = "earth"):
    if reference not in ["earth", "jupiter"]:
        raise ValueError("Ref mass must be 'earth' or 'jupiter'")
    return masses[planet] / masses[reference]
````

````python
print(to_planet_mass("earth", reference="jupiter"))

print(to_planet_mass("jupiter", reference="earth"))
````

**Q11) Write a function that returns two values.**

`mass_in_two_units(mass_1e24kg, reference)` should return the mass in Earth masses *and* in masses
of whatever reference planet you pass. Test it on Jupiter with `reference="mars"`.

Hint: `return a, b` returns a tuple, and `first, second = f(...)` unpacks it — both from 1.2's
lecture.

````python
# Q11: write mass_in_two_units here, then test it on Jupiter against mars
````

````python
def mass_in_two_units(planet, reference1="earth", reference2="mars"):

    in_earth_mass= masses[planet]/masses[reference1]
    in_planet_b_mass= masses[planet]/masses[reference2]
    return in_earth_mass, in_planet_b_mass
````

````python
jupiter_in_units = mass_in_two_units("jupiter", reference1="earth", reference2="mars")
print(jupiter_in_units)
````
